# 4. Visualização Computacional

**Ideia:** representar dados visualmente para facilitar sua compreensão.

**Aplicação escolhida:** gráfico de barras para comparar notas de alunos.

**Entrada:** nomes e notas. **Saída:** um gráfico.


```python
from pathlib import Path
import matplotlib.pyplot as plt

Path("resultados").mkdir(exist_ok=True)
alunos = ["Ana", "Bruno", "Carla"]
notas = [7, 9, 6]

fig, ax = plt.subplots(figsize=(6, 4))
barras = ax.bar(alunos, notas, color=["tab:blue", "tab:orange", "tab:green"])
ax.bar_label(barras, padding=3)
ax.set_title("Notas dos alunos — dados fictícios")
ax.set_xlabel("Aluno")
ax.set_ylabel("Nota")
ax.set_ylim(0, 10)
fig.tight_layout()
fig.savefig("resultados/4_visualizacao.png", dpi=150)
plt.show()
```

Cada barra representa um aluno; sua **altura representa a nota**. O eixo vertical vai de 0 a 10. Podemos perceber rapidamente que Bruno tem a maior nota: 9.

A visualização usa recursos de computação gráfica para **comunicar dados**. Neste exemplo, a altura das barras tem um significado numérico.
