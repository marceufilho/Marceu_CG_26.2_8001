# 2. Processamento de Imagens

**Ideia:** modificar uma imagem que já existe.

**Aplicação escolhida:** filtro de tons de cinza, como em um editor de fotos.

**Entrada:** imagem colorida. **Saída:** a mesma cena em tons de cinza.

Adaptação da conversão BGR → cinza explicada no [tutorial de cores do repositório OpenCV](https://github.com/opencv/opencv/blob/4.x/doc/py_tutorials/py_imgproc/py_colorspaces/py_colorspaces.markdown). Usamos uma imagem fixa para mostrar apenas essa operação.


```python
import cv2
from IPython.display import Image, display

imagem = cv2.imread("resultados/1_sintese.png")
assert imagem is not None, "Execute o notebook 01 para criar a imagem de entrada."

print("Antes: imagem colorida")
display(Image(filename="resultados/1_sintese.png"))
```

Cada pixel colorido possui três valores: azul, verde e vermelho. A função abaixo combina esses valores em uma intensidade de cinza.

```python
cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

cv2.imwrite("resultados/2_processamento.png", cinza)
print("Depois: imagem em tons de cinza")
display(Image(filename="resultados/2_processamento.png"))
```

As posições e os tamanhos dos círculos continuam iguais; seus pixels passam a representar intensidades de cinza. A saída continua sendo **uma imagem**.
