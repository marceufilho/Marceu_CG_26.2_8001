# 1. Síntese de Imagens (Computação Gráfica)

**Ideia:** criar uma imagem usando formas, posições e cores.

**Aplicação escolhida:** desenho 2D de três círculos, como em um programa de desenho.

**Entrada:** posição, raio e cor de cada círculo. **Saída:** uma imagem colorida.

Exemplo simplificado do [tutorial de desenho do repositório OpenCV](https://github.com/opencv/opencv/blob/4.x/doc/py_tutorials/py_gui/py_drawing_functions/py_drawing_functions.markdown). A adaptação desenha somente três círculos sobre um fundo branco.

Execute a célula com **Shift + Enter**. O resultado aparece abaixo.

```python
from pathlib import Path
import cv2
import numpy as np
from IPython.display import Image, display

Path("resultados").mkdir(exist_ok=True)

# Imagem branca: 300 pixels de altura, 600 de largura e 3 canais de cor.
imagem = np.full((300, 600, 3), 255, dtype=np.uint8)

# circle(imagem, centro, raio, cor, preenchimento)
# No OpenCV, a ordem das cores é BGR: azul, verde, vermelho.
cv2.circle(imagem, (100, 150), 50, (0, 0, 255), -1)
cv2.circle(imagem, (300, 150), 50, (0, 160, 0), -1)
cv2.circle(imagem, (500, 150), 50, (255, 0, 0), -1)

cv2.imwrite("resultados/1_sintese.png", imagem)
display(Image(filename="resultados/1_sintese.png"))
```

O computador transforma a descrição das formas em **pixels**, os pequenos pontos da imagem. Isso é uma forma simples de **renderização 2D**. O valor `50` indica o raio; `-1` manda preencher o círculo.

**Para apresentar:** “Aqui eu forneço formas e cores, e o computador cria uma imagem.”

**Experimente:** troque o raio do primeiro círculo de `50` para `30` e execute novamente. Depois volte para `50` para manter o exemplo original.
